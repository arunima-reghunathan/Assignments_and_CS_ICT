1 .Write a program that takes the user's account balance and withdrawal amount. If
sufficient balance is available, deduct the amount and display the updated balance.
Otherwise, print an error message.

Case Study:

1: Take Input --> Get the cuurent balance and amt to be withdrawn

2: check for sufficient balance and withdraw --> compare the withdrawal amt and balance with if, and condition satisfies withdraw from current balance

3:Withdraw and display ---> deduct the withdrawal amt and display the new balance
incase of insufficient balance show error msg

```python
# Get Input
Balance=float(input("Enter user balance:"))
Withdrawal=float(input("Enter amount to be withdrawn:"))

# Check for sufficient balance
if Withdrawal <= Balance:
  # withrdraw and display
  Balance = Balance-Withdrawal          #updating the new balance
  print("Transaction successful. Remaining balance:",Balance )      #display new balance
else:
  print("Transaction Failed , Insufficient Balance")                #Display error message
```

2.Create a bill calculator where the user enters the number of units consumed. The rate is
₹5 per unit for the first 100 units, ₹7 for the next 100, and ₹10 for anything above 200
units. Calculate and display the total bill.

case study:

1.take input --> units consumed

2.Calculate bill --> 0-100 - 5/unit,101-200 -7/unit, >200 - 10 /unit
first check and  calculate for units consumed <= 100 and then <= 200 and then above 200 using if and elif

3.Dispaly amt --> Printing final amount as per comsumption

```python
# Take input
units=int(input("Enter consumed units:"))

# Calculate bill
if units <= 100:                                                                 # if unit consumed is below first 100 charge 5 per unit
  Bill = units * 5
elif units <= 200:                                                               # if unit consumed is below first 100 charge 5 per unit and within next 100 units charge 7 unit
  Bill = 100 * 5 + (units-100) * 7
elif units >= 200:
  Bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)                            # same as above and charging 10 per unit for consumption above 200
else:
  print("Error occured, Try again")

# Display amt                                             # else print error msg
print(" Bill amount = ₹",Bill)
```

3. Write a Python program to convert a given temperature value between Celsius and
Fahrenheit. Accept input as a number and a unit indicator ('C' or 'F').


Case Study:
1. take input: get temperature value and unit as C or F
2. Check the units with if statement and use appropriate formulas to convert

```python
# take input
temp = float(input("Enter temperature:"))
unit = input("Enter unit as C or F:")

# convert temperature and display
if unit == "C" :
  f= (temp * 9/5) + 32  # formula to convert C to F
  print( "Celsius to Fahrenheit :", f,"°F")
elif unit  == "F" :
  c = (temp - 32) * 5/9 #formula to convert F to C
  print( "Fahrenheit to Celsius:",c,"°C")
else:
  print("Error, Try again !")
```

4. Write a program that takes the recharge date and today’s date as input and checks
whether the 28-day recharge plan is still valid or expired.

Case Study:

1. take input: get recharge date and todays date
2. convert strings to date object to find the numberof dates
3. Calculate expiry date


```python
# import from library
from datetime import datetime, timedelta

# Read input dates
recharge_date_str = input("Enter recharge date (YYYY-MM-DD): ")
today_date_str = input("Enter today's date (YYYY-MM-DD): ")

# Convert strings to date objects
recharge_date = datetime.strptime(recharge_date_str, "%Y-%m-%d").date()
today_date = datetime.strptime(today_date_str, "%Y-%m-%d").date()

# Calculate expiry date (28 days plan)
expiry_date = recharge_date + timedelta(days=28)

# Check validity
if today_date <= expiry_date:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")
```

7. Write a program that takes input of 3 students’ names and their marks, stores them in a
dictionary, and prints the average marks.

Case Study

1. Create empty dictionary
2. Set number of students as 3
3. create loop --> take input --> name and marks --> store as key value pair
4. Calculate total and average marks

```python
students = {}
num = 3
# Loop to take input for 3 students
for i in range(num):
    name = input("Enter name : ")
    marks = float(input("Enter marks: "))
    students[name] = marks  # store in dictionary

# Calculate average
Total_marks = sum(students.values())

average = Total_marks / num


# Display result
print(f"Average Marks: {average}")
```

8. Write a program that takes a sentence and displays the frequency of each word using a
dictionary.

Case Study
1. Take input sentence
2. Split it into separate words
3. create loop --> for word in word list , add to freq list if it does not exist, if it exists add +1 to count
4. display results

```python
# Input sentence from user
sentence = input("Enter a sentence: ")

# Split sentence into  list of words
word_list = sentence.split()

# Initialize empty dictionary to store word frequencies
frequency = {}

# Count frequency of each word
for word in word_list:
    if word in frequency:
        frequency[word] += 1  # increment count if word already exists
    else:
        frequency[word] = 1   # add word to dictionary if not exists

# Display word frequencies
for word, count in frequency.items():
    print(f"{word}: {count}")
```

9. Write a function that checks whether a number is prime. Call the function with user input

Case Study
1. define function to check prime
2. set is_prime a true and create loop if number divisible by any number --> not prime
3. Call function with input number as argument

```python
# Function to check if a number is prime
def check_prime(num):
    """ To check if the given number is a prime number """
    if num <= 1:
        print(num," is not a prime number.")
        return

    # Assume the number is prime
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
       print(num,"is a prime number.")
    else:
        print(num," is not a prime number.")

# Input from user
number = int(input("Enter a number: "))

# Call the function
check_prime(number)
```

10. Write a function that takes a string and returns the number of vowels in it.

Case study
1. Define function --> define set of vowels -- for every char in sentence in vowels count +1
2. take input sentence
3. call the function

```python
# define function to count vowels
def count_vowels(sentence):
    """ To count vowels """
    vowels = "aeiouAEIOU" # all vowels (lowercase + uppercase)
    count = 0             # initialize counter

    for char in sentence:
        if char in vowels:
            count = count + 1    # increment count if character is a vowel

    print("number of vowels are :",count)

# Input sentence from user
sentence = input("Enter a sentence: ")

# Call the function
count_vowels(sentence)
```

11. Write a program that takes two numbers as input: a range limit N and a divisor D. Count
how many numbers between 1 and N are divisible by D.

Case Study
1. Take inputs from user
2. create loop --> for numbers in range 1 to N -- Check if it is divisible by D if divisible count + 1
3. display results

```python
# Take input from user
N = int(input("Enter the range limit (N): "))
D = int(input("Enter the divisor (D): "))

count = 0  # Initialize count

# Loop through numbers from 1 to N
for i in range(1, N+1):
    if i % D == 0:  # check if number is divisible by D
        count = count + 1

print("Count of numbers divisible by",D,":", count)
```
