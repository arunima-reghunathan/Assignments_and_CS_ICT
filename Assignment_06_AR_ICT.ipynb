Assignment 06 - SENTIMENT ANALYSIS

Natural language processing has extensively applied in sentiment analysis as
a significant challenge. In this scenario, the objective is to determine if the
tweets shared by customers regarding technology companies that produce
and sell mobiles, computers, laptops, and similar products express positive
sentiment or negative sentiment. The goal will be to build a system that can
accurately classify the new tweets sentiments. You can divide the data into
train and test. The Evaluation metric you should use is the accuracy.


```python
# load dataset
import pandas as pd
import io
df = pd.read_csv("/content/tweets.csv")
```

```python
# display data
df
```

```python
# Import libraries
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
nltk.download('wordnet')
nltk.download('stopwords')
```

Clean Text

```python
# clean tweets
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r"http\S+|@\w+|#\w+", "", text)   # remove URLs, mentions
    text = re.sub(r"[^a-z\s]", "", text)            # remove punctuation

    words = text.split()
    words = [lemmatizer.lemmatize(w) for w in words if w not in stop_words]

    return " ".join(words)

df['tweet'] = df['tweet'].apply(preprocess)
```

```python
df
```

TF-IDF

```python
vectorizer = TfidfVectorizer(max_features=5000)

X = vectorizer.fit_transform(df['tweet'])
y = df['label']
```

Split data into training and testing

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

Build Model

```python
# Logistic Regression model
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
#Prediction
y_pred = model.predict(X_test)
```

Evaluation

```python
# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
#Classification Report
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
#Confusion Matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

The Logistic Regression model achieved an accuracy of 83.33%, indicating that it can correctly classify most tweets as positive or negative.
